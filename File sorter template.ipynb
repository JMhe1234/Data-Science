```python
import os, shutil
```

```python
#when copying the file path, change to / and add one at the end so the code knows to create files within a folder and not outside on the main e.g.
#do the sorting by creating files within another folder in docuemnts not create the new folders in documents 
path = r"add file path that has file that need sorting"
```

```python
file_name = os.listdir(path)
```

```python
#create list for all folder needed
folder_name = ['Excel Worksheet File','Text File','Image File']

#fucntion to create sorter
def file_sorter():
    #checking if folder exists, if not will create new folder in assigned path
    for loop in range(0,3):
        if not os.path.exists(path + folder_name[loop]):
            #print(path + folder_name[loop])
            os.makedirs((path + folder_name[loop]))

    #moving the files based on file type into correct folder, will differ to the types on files that need moving and whihc folder, I used test files
    for file in file_name:
        if ".xlsx" in file and not os.path.exists(path + "Excel Worksheet File/" + file):
            shutil.move(path + file, path + "Excel Worksheet File/" + file)
        elif ".txt" in file and not os.path.exists(path + "Text File/" + file):
            shutil.move(path + file, path + "Text File/" + file)
        elif ".jpg" in file and not os.path.exists(path + "Image File/" + file):
            shutil.move(path + file, path + "Image File/" + file)
```

```python
#calling fucntion
file_sorter()
```
